# 03 — Feature Engineering

Finalize the feature matrix. Inspect each feature for quality and document decisions.

**Output:** `data/processed/feature_matrix.parquet`

## Setup

```
import sys
sys.path.insert(0, '..')
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
%config InlineBackend.figure_format = 'retina'
```

## 1. Build Feature Matrix

```
from features.build import build_feature_matrix, save_feature_matrix, FEATURE_COLUMNS, LABEL_COLUMN

df = build_feature_matrix(use_synthetic_tidal=True)
print(df[FEATURE_COLUMNS + [LABEL_COLUMN]].describe())
```

## 2. Check for Missing Values

```
missing = df[FEATURE_COLUMNS].isnull().sum()
print("Missing values per feature:")
print(missing)
print()
if missing.any():
    print("⚠️  Missing values detected. Document imputation strategy below.")
else:
    print("✅ No missing values in feature columns.")
```

## 3. Feature Scaling Preview

Features will be standardized (zero mean, unit variance) inside the model training loop. Do not scale here — scaling must happen within each CV fold to avoid leakage.

```
from sklearn.preprocessing import StandardScaler
scaler_preview = StandardScaler()
X = df[FEATURE_COLUMNS].values
X_scaled = scaler_preview.fit_transform(X)

print("After standardization (preview only — not saved):")
print(f"Mean: {X_scaled.mean(axis=0).round(4)}")
print(f"Std:  {X_scaled.std(axis=0).round(4)}")
```

## 4. Spatial Maps of Each Feature

```
fig, axes = plt.subplots(2, 2, figsize=(16, 9))

for ax, col in zip(axes.ravel(), FEATURE_COLUMNS):
    sc = ax.scatter(df['lon_centre'], df['lat_centre'],
                    c=df[col], cmap='viridis', s=0.5, alpha=0.8)
    plt.colorbar(sc, ax=ax, shrink=0.8)
    ax.set_title(col, fontweight='bold')
    ax.set_xlim(-180, 180); ax.set_ylim(-90, 90)
    ax.set_xlabel('Longitude'); ax.set_ylabel('Latitude')

plt.suptitle('Spatial Distribution of Features', fontweight='bold')
plt.tight_layout()
plt.show()
```

## 5. Save Final Feature Matrix

```
save_feature_matrix(df)
print("✅ Feature matrix saved to data/processed/feature_matrix.parquet")
print(f"   Shape: {df.shape}")
print(f"   Features: {FEATURE_COLUMNS}")
print(f"   Label: {LABEL_COLUMN}")
```
